## floor, room_image, room_type 컬럼 추가

```python
adf4 = adf4.assign(floor=np.nan,
                   room_image=np.nan,
                   room_type=np.nan)
adf4
```

## floor 추가

안내실과 7417-B가 포함된 행은 예외로 나중에 직접 수정해준다.

```python

def extract_floor(val):
    # 결측치이거나 문자열 변환 후 길이가 부족한 경우 처리
    if pd.isna(val):
        return None

    text = str(val)
    # 길이가 5 이상이면 -> 3번째 글자 (인덱스 2)
    if len(text) >= 5:
        return text[2]
    # 길이가 5 미만 -> 2번째 글자 (인덱스 1)
    elif len(text) >= 2:
        return text[1]
    else:
        return None

# floor 추가
adf4['floor'] = adf4['room'].apply(extract_floor)
```

```python
# 1. 특정 텍스트가 포함된 조건을 마스크(mask)로 정의
mask = adf4['room'].str.contains('7417-B', na=False)
# 2. loc[행_조건, '수정할컬럼명'] = 변경할_값
adf4.loc[mask, 'floor'] = '4'

mask = adf4['room'].str.contains('경비', na=False)
adf4.loc[mask, 'floor'] = '1'
```
