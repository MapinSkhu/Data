# 필요한 컬럼만 남기기

1. 관, 층, 세부 만 남길 것이다. (kwan, room, detail)

2. adf에서 사용하는 관 이름이 맵인스쿠에서 사용하는 이름과 다르기에 맵인스쿠 이름과 동일하게 바꿔줄 것이다.

3. 세부는 해당 강의실에 대한 정보가 담겨 있기에 층(세부) 형식으로 사용하고 버릴 것이다.

4. 행복기숙사 행은 버릴 것이다. (인덱스 387 ~ 397)

5. 맵인스쿠에서 강의가 없는 강의실은 검색을 해야만 확인이 가능하다.  강의가 없는 강의실도 검색하지 않아도 보일 수 있게 바꿀지 생각해보아야 한다.

```python
adf2 = adf[['room', 'detail']]
adf2.room = adf2.room + '(' + adf2.detail + ')'
adf2 = adf2.drop('detail', axis=1)
adf2.room = adf2.room.str.replace(' ', '')
adf2
```
